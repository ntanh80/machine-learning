# Đạo hàm riêng của hàm mất mát MSE

*Bài 5, mục 5.1.2.* Dữ liệu 3 điểm $(1, 2)$, $(2, 3)$, $(3, 5)$, mô hình $\hat{y} = w_0 + w_1 x$:

$$\mathcal{L}(\mathbf{w}) = \frac{1}{2N}\|\mathbf{X}\mathbf{w} - \mathbf{y}\|^2, \qquad \frac{\partial \mathcal{L}}{\partial w_0} = \frac{1}{N}\sum_i e_i, \qquad \frac{\partial \mathcal{L}}{\partial w_1} = \frac{1}{N}\sum_i e_i x_i, \qquad e_i = \hat{y}_i - y_i$$

```python
import numpy as np

X = np.array([[1, 1], [1, 2], [1, 3]], dtype=float)   # mỗi hàng [1, x_i]
y = np.array([2, 3, 5], dtype=float)


def loss(w):
    return np.sum((X @ w - y)**2) / (2 * len(y))


def grad(w):
    # (1/N) X^T (Xw - y): thành phần j là (1/N) sum(e_i * x_ij)
    return X.T @ (X @ w - y) / len(y)
```

## 1. Tính từng đạo hàm riêng tại $\mathbf{w} = (0, 0)$

```python
w = np.zeros(2)
e = X @ w - y
print('Sai số e          :', e)
print('L(0, 0)           : %.3f' % loss(w))
print('dL/dw0 = mean(e)  : %.3f' % e.mean())
print('dL/dw1 = mean(e*x): %.3f' % np.mean(e * X[:, 1]))
print('grad (công thức)  :', grad(w))
```

## 2. Kiểm tra gradient bằng đạo hàm số

$$\frac{\partial \mathcal{L}}{\partial w_j} \approx \frac{\mathcal{L}(\mathbf{w} + \varepsilon\mathbf{e}_j) - \mathcal{L}(\mathbf{w} - \varepsilon\mathbf{e}_j)}{2\varepsilon}$$

```python
def numerical_grad(w, eps=1e-6):
    g = np.zeros_like(w)
    for j in range(len(w)):
        e = np.zeros_like(w); e[j] = eps
        g[j] = (loss(w + e) - loss(w - e)) / (2 * eps)
    return g


print('grad (đạo hàm số):', numerical_grad(w))
print('Khớp công thức   :', np.allclose(grad(w), numerical_grad(w)))
```
