# Perceptron với scikit-learn

*Bài 4, mục 4.2.3 e).* So sánh `Perceptron` và `LogisticRegression` trên cùng dữ liệu ở notebook 07.

```python
import numpy as np
from sklearn.linear_model import Perceptron, LogisticRegression

np.random.seed(2)
means = [[-1, 0], [1, 0]]
cov = [[.3, .2], [.2, .3]]
N = 10
X0 = np.random.multivariate_normal(means[0], cov, N)   # lớp +1
X1 = np.random.multivariate_normal(means[1], cov, N)   # lớp -1
X = np.concatenate((X0, X1), axis=0)
y = np.concatenate((np.ones(N), -1 * np.ones(N)))
```

```python
perceptron_model = Perceptron(max_iter=1000, random_state=0).fit(X, y)
print('Perceptron: w0 =', perceptron_model.intercept_, ' w =', perceptron_model.coef_, ' Accuracy =', perceptron_model.score(X, y))
```

Hồi quy Logistic cũng có ranh giới tuyến tính nhưng cho thêm **xác suất** dự đoán; Perceptron chỉ cho nhãn cứng.

```python
logistic_model = LogisticRegression().fit(X, y)
print('Logistic  : w0 =', logistic_model.intercept_, ' w =', logistic_model.coef_, ' Accuracy =', logistic_model.score(X, y))
print('Xác suất lớp +1 của 3 điểm đầu:', logistic_model.predict_proba(X[:3])[:, 1].round(3))
```
