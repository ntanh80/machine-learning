# Pandas: đọc và khám phá dữ liệu dạng bảng

Pandas cung cấp kiểu `DataFrame`: bảng dữ liệu có tên cột, giống một trang tính Excel. Dữ liệu [../data/students.csv](../data/students.csv) gồm 20 sinh viên:

| Cột | Ý nghĩa |
|---|---|
| `study_hours` | Số giờ tự học mỗi tuần |
| `attendance` | Số buổi đi học (trên 15 buổi) |
| `midterm` | Điểm giữa kỳ (thang 10) |
| `passed` | Kết quả học phần: 1 = đạt, 0 = không đạt |

## 1. Đọc dữ liệu

```python
from pathlib import Path
import pandas as pd

DATA = Path('..') / 'data'                      # thư mục dữ liệu của bài
df = pd.read_csv(DATA / 'students.csv')

print('Kích thước:', df.shape)                  # (số dòng, số cột)
df.head()                                       # 5 dòng đầu
```

## 2. Thông tin tổng quan

```python
df.info()                                       # kiểu dữ liệu, số giá trị khác rỗng của từng cột
```

```python
df.describe().round(2)                          # thống kê mô tả các cột số
```

## 3. Chọn cột, lọc dòng

```python
print(df['midterm'].mean())                     # một cột là một Series
df[['study_hours', 'midterm']].head(3)           # nhiều cột là một DataFrame
```

```python
df[df['passed'] == 1].head()                    # các sinh viên đạt
```

## 4. Thống kê theo nhóm

`groupby` gom các dòng theo giá trị của một cột rồi tính thống kê cho từng nhóm. Kết quả cho thấy nhóm đạt có số giờ học, số buổi đi học và điểm giữa kỳ trung bình cao hơn.

```python
print(df['passed'].value_counts())              # số sinh viên mỗi nhóm
df.groupby('passed').mean().round(2)              # trung bình từng cột theo nhóm
```

## 5. Tạo cột mới và tách đặc trưng, nhãn

Trước khi đưa vào mô hình, ta tách bảng thành ma trận đặc trưng `X` và vector nhãn `y`.

```python
df['attendance_rate'] = df['attendance'] / 15        # tỉ lệ đi học

X = df[['study_hours', 'attendance', 'midterm']].values   # ma trận đặc trưng (mảng NumPy)
y = df['passed'].values                                 # vector nhãn
print('X:', X.shape, ' y:', y.shape)
```
