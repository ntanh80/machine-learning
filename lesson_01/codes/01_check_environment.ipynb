# Kiểm tra môi trường Anaconda

Chạy notebook này sau khi cài Anaconda và tạo môi trường `ml` (mục 1.2.1). Nếu mọi ô chạy không lỗi và in ra phiên bản các thư viện, môi trường đã sẵn sàng cho cả học phần.

## 1. Phiên bản Python và thư mục cài đặt

`sys.executable` cho biết notebook đang dùng Python nào. Đường dẫn phải nằm trong thư mục của Anaconda (ví dụ `...\anaconda3\envs\ml\python.exe`); nếu không, kernel đang chọn sai môi trường.

```python
import sys

print('Phiên bản Python:', sys.version.split()[0])
print('Đường dẫn Python:', sys.executable)
```

## 2. Phiên bản các thư viện dùng trong học phần

```python
import numpy as np
import pandas as pd
import matplotlib
import sklearn

libraries = {'NumPy': np, 'Pandas': pd, 'Matplotlib': matplotlib, 'Scikit-learn': sklearn}
for name, module in libraries.items():
    print(f'{name:<13}: {module.__version__}')
```
