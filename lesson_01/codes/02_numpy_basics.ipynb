# NumPy: mảng và phép toán trên mảng

NumPy cung cấp kiểu dữ liệu **mảng nhiều chiều** `ndarray` và các phép toán nhanh trên mảng. Trong học máy, vector đặc trưng $\mathbf{x}$ là mảng 1 chiều, ma trận dữ liệu $\mathbf{X}$ là mảng 2 chiều.

## 1. Tạo mảng

```python
import numpy as np

x = np.array([170, 65, 22])        # vector đặc trưng của một người: chiều cao, cân nặng, tuổi
X = np.array([[170, 65, 22],       # ma trận dữ liệu: mỗi hàng là một mẫu
              [158, 50, 20],
              [165, 58, 21]])

print('x =', x, '- số chiều:', x.ndim, '- kích thước:', x.shape)
print('X có kích thước', X.shape, '(3 mẫu, 3 đặc trưng)')
```

```python
print(np.zeros(3))                  # mảng toàn số 0
print(np.ones((2, 3)))              # ma trận 2 x 3 toàn số 1
print(np.arange(0, 10, 2))          # dãy số từ 0 đến 10 (không gồm 10), bước 2
print(np.linspace(0, 1, 5))         # 5 số cách đều trong đoạn [0, 1]
```

## 2. Truy cập phần tử

`X[i, j]` là phần tử hàng `i`, cột `j` (đánh số từ 0). Dấu `:` nghĩa là lấy tất cả.

```python
print('Mẫu đầu tiên    :', X[0])        # hàng 0
print('Cột chiều cao   :', X[:, 0])     # cột 0 của mọi hàng
print('Cân nặng mẫu thứ 2:', X[1, 1])
print('Các mẫu cao trên 160 cm:\n', X[X[:, 0] > 160])   # lọc theo điều kiện
```

## 3. Phép toán trên mảng (vector hóa)

Phép toán được áp dụng cho **từng phần tử** mà không cần vòng lặp `for`. Đây là cách viết code học máy nhanh và gọn.

```python
height_m = X[:, 0] / 100                     # đổi cm sang m cho cả cột
bmi = X[:, 1] / height_m**2                  # chỉ số BMI = cân nặng / chiều cao^2
print('BMI:', bmi.round(1))

print('Trung bình từng cột:', X.mean(axis=0).round(2))   # axis=0: tính theo cột
print('Lớn nhất từng cột  :', X.max(axis=0))
```

## 4. Nhân ma trận

Dự đoán của mô hình tuyến tính cho cả tập dữ liệu là $\hat{\mathbf{y}} = \mathbf{X}\mathbf{w}$ (học ở Bài 3). Trong NumPy viết `X @ w`.

```python
w = np.array([0.5, 0.2, -0.1])   # vector trọng số w (ví dụ)
y_hat = X @ w                    # mỗi phần tử là tổng có trọng số của một hàng
print('X @ w =', y_hat)
print('X.T có kích thước', X.T.shape)   # ma trận chuyển vị
```
