# Biểu diễn văn bản bằng túi từ (Bag of Words)

Mỗi câu được biểu diễn bằng một vector, mỗi phần tử là số lần xuất hiện của một từ trong từ điển. Dữ liệu viết thường, không có dấu câu để dễ tách bằng `split()`.

```python
from sklearn.feature_extraction.text import CountVectorizer

van_ban = [
    'tôi thích học máy',
    'tôi thích python',
    'học máy học python',
]
```

## Cách 1: Python thuần

Tách theo khoảng trắng (tiếng Việt khi đó được tách thành các âm tiết).

```python
cac_cau = [cau.split() for cau in van_ban]
tu_dien = sorted(set(' '.join(van_ban).split()))

ma_tran = []
for cau in cac_cau:
    hang = []
    for tu in tu_dien:
        hang.append(cau.count(tu))
    ma_tran.append(hang)

print('Từ điển:', tu_dien)
for hang in ma_tran:
    print(hang)
```

## Cách 2: scikit-learn

Mỗi hàng là một câu, mỗi cột là số lần xuất hiện của một từ.

```python
bo_dem = CountVectorizer()
ma_tran_sklearn = bo_dem.fit_transform(van_ban).toarray()

print('Từ điển:', bo_dem.get_feature_names_out().tolist())
print(ma_tran_sklearn)
```

## Cách 3: TF-IDF

Trọng số từ thay cho số lần xuất hiện: từ có trong ít câu được tăng trọng số; mỗi vector được chuẩn hóa L2.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
ma_tran_tfidf = tfidf.fit_transform(van_ban).toarray()

print('Từ điển:', tfidf.get_feature_names_out().tolist())
print(ma_tran_tfidf.round(3))
```

## Cách 4: Hashing

Băm từ vào vector cố định 8 chiều. Có thể có nhiều từ chung một cột; đây là biểu diễn bằng băm, không phải embedding học ngữ nghĩa.

```python
from sklearn.feature_extraction.text import HashingVectorizer

hashing = HashingVectorizer(n_features=8, alternate_sign=False)
ma_tran_hash = hashing.transform(van_ban).toarray()
print(ma_tran_hash.round(3))
```

## Cách 5: LSA

Dùng SVD nén TF-IDF thành embedding 2 chiều.

```python
from sklearn.decomposition import TruncatedSVD

lsa = TruncatedSVD(n_components=2, random_state=42)
embedding_lsa = lsa.fit_transform(ma_tran_tfidf)
print(embedding_lsa.round(3))
```

## Cách 6: NMF

Biểu diễn mỗi câu bằng 2 trọng số chủ đề không âm.

```python
from sklearn.decomposition import NMF

nmf = NMF(n_components=2, init='nndsvda', random_state=42, max_iter=500)
embedding_nmf = nmf.fit_transform(ma_tran_tfidf)
print(embedding_nmf.round(3))
```

Mỗi hàng vẫn là một câu; cột LSA/NMF là đặc trưng ẩn, không phải từ. Chỉ có 3 câu nên đây là minh họa cách chạy, chưa học ngữ nghĩa tốt.
