# Sinh dữ liệu mẫu 10000 căn nhà

Dùng cho bài toán dự đoán giá nhà (Bài 03 – Hồi quy tuyến tính).

| Ký hiệu | Đặc trưng | Cột dữ liệu | Đơn vị |
|---|---|---|---|
| $x_1$ | Diện tích | `area` | m² |
| $x_2$ | Số phòng ngủ | `bedrooms` | phòng |
| $x_3$ | Khoảng cách tới trung tâm | `distance` | km |
| $y$ | Giá nhà | `price` | triệu VND |

Giá nhà tăng theo diện tích, số phòng ngủ và giảm khi xa trung tâm:

$$y = 500 + 30x_1 + 150x_2 - 60x_3 + \text{nhiễu}$$

**Lưu ý:** chạy notebook này sẽ ghi đè file `house_prices.csv`.

```python
import numpy as np
import pandas as pd

np.random.seed(42)   # cố định bộ sinh số ngẫu nhiên để lần chạy nào cũng ra cùng dữ liệu
N = 10000            # số căn nhà
```

```python
area = np.round(np.random.uniform(30, 200, N), 1)          # diện tích x1 (m²)
# Số phòng ngủ x2 phụ thuộc vào diện tích (nhà rộng thường nhiều phòng hơn)
bedrooms = np.clip(np.round(area / 40 + np.random.normal(0, 0.7, N)), 1, 6).astype(int)
distance = np.round(np.random.uniform(0.5, 25, N), 1)      # khoảng cách x3 (km)

noise = np.random.normal(0, 300, N)                        # nhiễu ngẫu nhiên
price = 500 + 30 * area + 150 * bedrooms - 60 * distance + noise   # giá nhà y (triệu VND)
price = np.round(np.maximum(price, 300), 0)                # giá thấp nhất 300 triệu

df = pd.DataFrame({
    'area': area,
    'bedrooms': bedrooms,
    'distance': distance,
    'price': price,
})
```

```python
df.to_csv('house_prices.csv', index=False)
print('Đã lưu', len(df), 'mẫu vào house_prices.csv')
df.describe()
```
