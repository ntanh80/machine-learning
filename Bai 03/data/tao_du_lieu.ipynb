# Sinh dữ liệu mẫu 1000 căn nhà

Dùng cho bài toán dự đoán giá nhà (Bài 03 - Hồi quy tuyến tính).

- $x_1$: diện tích (m²)
- $x_2$: số phòng ngủ (phòng)
- $x_3$: khoảng cách tới trung tâm (km)
- $y$: giá nhà (triệu VND)

Giá nhà tăng theo diện tích, số phòng ngủ và giảm khi xa trung tâm:

$$y = 500 + 30x_1 + 150x_2 - 60x_3 + 	ext{nhiễu}$$

**Lưu ý:** chạy notebook này sẽ ghi đè file `gia_nha.csv`.

```python
import numpy as np
import pandas as pd

np.random.seed(42)
N = 1000
```

```python
dien_tich = np.round(np.random.uniform(30, 200, N), 1)
# Số phòng ngủ phụ thuộc vào diện tích (nhà rộng thường nhiều phòng hơn)
so_phong_ngu = np.clip(np.round(dien_tich / 40 + np.random.normal(0, 0.7, N)), 1, 6).astype(int)
khoang_cach = np.round(np.random.uniform(0.5, 25, N), 1)

nhieu = np.random.normal(0, 300, N)
gia = 500 + 30 * dien_tich + 150 * so_phong_ngu - 60 * khoang_cach + nhieu
gia = np.round(np.maximum(gia, 300), 0)

df = pd.DataFrame({
    'dien_tich': dien_tich,
    'so_phong_ngu': so_phong_ngu,
    'khoang_cach': khoang_cach,
    'gia': gia,
})
```

```python
df.to_csv('gia_nha.csv', index=False)
print('Đã lưu', len(df), 'mẫu vào gia_nha.csv')
df.describe()
```
