# min, max, argmin và argmax

- `min` / `max` trả về **giá trị** nhỏ nhất / lớn nhất.
- `argmin` / `argmax` trả về **vị trí (chỉ số)** của giá trị đó.

Trong học máy:

- Huấn luyện mô hình: $w^* = \arg\min_w L(w)$, tức tìm tham số làm hàm mất mát nhỏ nhất.
- Phân loại: $\hat{y} = \arg\max_k P(y = k \mid x)$, tức chọn lớp có xác suất cao nhất.

```python
import numpy as np
```

# Phần 1: Hàm min và max

## Hàm có sẵn của Python

`min` / `max` nhận một danh sách (hoặc nhiều đối số) và trả về giá trị nhỏ nhất / lớn nhất.

```python
diem = [7.5, 9.0, 6.0, 8.5, 6.0]

print("Điểm:", diem)
print("min(diem) =", min(diem))
print("max(diem) =", max(diem))

# Truyền trực tiếp nhiều đối số
print("min(4, 1, 7) =", min(4, 1, 7))
print("max(4, 1, 7) =", max(4, 1, 7))
```

## Hàm min, max của NumPy

Với mảng NumPy, dùng `np.min` / `np.max` (hoặc `a.min()` / `a.max()`), nhanh hơn và hỗ trợ tham số `axis`.

```python
a = np.array(diem)
print("np.min(a) =", np.min(a), "| a.min() =", a.min())
print("np.max(a) =", np.max(a), "| a.max() =", a.max())
```

# Phần 2: Hàm argmin và argmax

## Cách 1: NumPy với mảng 1 chiều

```python
diem = [7.5, 9.0, 6.0, 8.5, 6.0]
a = np.array(diem)

print(a)
print("np.argmin(a) =", np.argmin(a))
print("np.argmax(a) =", np.argmax(a))
```

**Lưu ý:** có hai giá trị 6.0 bằng nhau (vị trí 2 và 4), `np.argmin` lấy vị trí xuất hiện **đầu tiên**.
