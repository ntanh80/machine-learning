# Phần 1: Vector $\mathbf{x}$ và chuyển vị $\mathbf{x}^T$

Quy ước trong bài Hồi quy tuyến tính: vector là **vector cột**.

$$
\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix} = [x_1, x_2, x_3]^T,
\qquad
\mathbf{x}^T = [x_1, x_2, x_3]
$$

Chuyển vị đổi hàng thành cột: $\mathbf{x}$ có kích thước $3 \times 1$, còn $\mathbf{x}^T$ có kích thước $1 \times 3$.

## 1. Vector cột trong NumPy

Vector cột $\mathbf{x}$ được lưu bằng mảng 2 chiều có 3 hàng, 1 cột.

```python
import numpy as np

x = np.array([[1], [2], [3]])   # vector cột x, kích thước 3 x 1
print("x (vector cột):")
print(x)
print("shape:", x.shape)
```

## 2. Mảng 1 chiều không phải vector cột

Mảng `np.array([1, 2, 3])` chỉ có **1 chiều** (`shape = (3,)`), không phân biệt hàng hay cột, nên chuyển vị `.T` không làm thay đổi gì. Muốn có vector cột thì dùng mảng 2 chiều như ở mục 1, hoặc `reshape(-1, 1)`.

```python
a = np.array([1, 2, 3])          # mảng 1 chiều
print("a.shape   =", a.shape)
print("a.T.shape =", a.T.shape)  # vẫn là (3,)

column = a.reshape(-1, 1)        # đổi thành vector cột 3 x 1
print("a.reshape(-1, 1).shape =", column.shape)
```

## 3. Chuyển vector cột thành vector hàng

$$
\mathbf{x} = \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}
\quad \xrightarrow{\ T\ } \quad
\mathbf{x}^T = \begin{bmatrix} 1 & 2 & 3 \end{bmatrix}
$$

Kích thước đổi từ $3 \times 1$ thành $1 \times 3$.

```python
xT = x.T                         # chuyển vị x^T
print("x^T (vector hàng):")
print(xT)
print("shape:", xT.shape)
```

## 4. Chuyển vector hàng thành vector cột

$$
\mathbf{v} = \begin{bmatrix} 4 & 5 & 6 \end{bmatrix}
\quad \xrightarrow{\ T\ } \quad
\mathbf{v}^T = \begin{bmatrix} 4 \\ 5 \\ 6 \end{bmatrix}
$$

Kích thước đổi từ $1 \times 3$ thành $3 \times 1$.

```python
v = np.array([[4, 5, 6]])        # vector hàng v, kích thước 1 x 3
print("v (vector hàng):")
print(v)
print("shape:", v.shape)

vT = v.T                         # chuyển vị v^T
print("\nv^T (vector cột):")
print(vT)
print("shape:", vT.shape)
```

## 5. Chuyển vị hai lần thì được lại vector ban đầu

$(\mathbf{x}^T)^T = \mathbf{x}$: cột → hàng → cột, và hàng → cột → hàng.

```python
print("(x^T)^T == x:", np.array_equal(x.T.T, x))
print("(v^T)^T == v:", np.array_equal(v.T.T, v))
```

## min, max, argmin và argmax

- `min` / `max` trả về **giá trị** nhỏ nhất / lớn nhất.
- `argmin` / `argmax` trả về **vị trí (chỉ số)** của giá trị đó.

Trong học máy:

- Huấn luyện mô hình: $w^* = \arg\min_w L(w)$, tức tìm tham số làm hàm mất mát nhỏ nhất.
- Phân loại: $\hat{y} = \arg\max_k P(y = k \mid x)$, tức chọn lớp có xác suất cao nhất.

# Phần 2: Hàm min và max

## Hàm có sẵn của Python

`min` / `max` nhận một danh sách (hoặc nhiều đối số) và trả về giá trị nhỏ nhất / lớn nhất.

```python
scores = [7.5, 9.0, 6.0, 8.5, 6.0]   # điểm của 5 sinh viên

print("Điểm:", scores)
print("min(scores) =", min(scores))
print("max(scores) =", max(scores))

# Truyền trực tiếp nhiều đối số
print("min(4, 1, 7) =", min(4, 1, 7))
print("max(4, 1, 7) =", max(4, 1, 7))
```

## Hàm min, max của NumPy

Với mảng NumPy, dùng `np.min` / `np.max` (hoặc `a.min()` / `a.max()`), nhanh hơn và hỗ trợ tham số `axis`.

```python
a = np.array(scores)   # chuyển danh sách thành mảng NumPy
print("np.min(a) =", np.min(a), "| a.min() =", a.min())
print("np.max(a) =", np.max(a), "| a.max() =", a.max())
```

# Phần 3: Hàm argmin và argmax

## Cách 1: NumPy với mảng 1 chiều

```python
scores = [7.5, 9.0, 6.0, 8.5, 6.0]
a = np.array(scores)

print(a)
print("np.argmin(a) =", np.argmin(a))   # vị trí của điểm thấp nhất
print("np.argmax(a) =", np.argmax(a))   # vị trí của điểm cao nhất
```

**Lưu ý:** có hai giá trị 6.0 bằng nhau (vị trí 2 và 4), `np.argmin` lấy vị trí xuất hiện **đầu tiên**.
