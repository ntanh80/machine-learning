# Vector $\mathbf{x}$ và chuyển vị $\mathbf{x}^T$

Quy ước trong bài Hồi quy tuyến tính: vector là **vector cột**.

$$
\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix} = [x_1, x_2, x_3]^T,
\qquad
\mathbf{x}^T = [x_1, x_2, x_3]
$$

Chuyển vị đổi hàng thành cột: $\mathbf{x}$ có kích thước $3 \times 1$, còn $\mathbf{x}^T$ có kích thước $1 \times 3$.

```python
import numpy as np
```

## 1. Mảng 1 chiều trong NumPy không có "hàng" hay "cột"

Với mảng 1 chiều, `x.T` trả về đúng mảng cũ, không có gì thay đổi.

```python
x = np.array([1, 2, 3])
print("x =", x, "| shape:", x.shape)
print("x.T =", x.T, "| shape:", x.T.shape)
```

## 2. Tạo vector cột và vector hàng

Muốn phân biệt hàng và cột, phải dùng **mảng 2 chiều** (có 2 cặp ngoặc vuông):

- Vector cột: 3 hàng, 1 cột, kích thước $3 \times 1$
- Vector hàng: 1 hàng, 3 cột, kích thước $1 \times 3$

```python
cot = np.array([[1],
                [2],
                [3]])
print("Vector cột:")
print(cot)
print("shape:", cot.shape)

hang = np.array([[1, 2, 3]])
print("\nVector hàng:")
print(hang)
print("shape:", hang.shape)
```

Cũng có thể tạo từ mảng 1 chiều bằng `reshape`: `reshape(-1, 1)` cho vector cột, `reshape(1, -1)` cho vector hàng.

```python
a = np.array([1, 2, 3])
print("a.reshape(-1, 1) có shape:", a.reshape(-1, 1).shape)
print("a.reshape(1, -1) có shape:", a.reshape(1, -1).shape)
```

## 3. Chuyển vector cột thành vector hàng

$$
\mathbf{x} = \begin{bmatrix} 1 \ 2 \ 3 \end{bmatrix}
\quad \xrightarrow{\ T\ } \quad
\mathbf{x}^T = \begin{bmatrix} 1 & 2 & 3 \end{bmatrix}
$$

Kích thước đổi từ $3 \times 1$ thành $1 \times 3$.

```python
x = np.array([[1],
              [2],
              [3]])
print("x (vector cột):")
print(x)
print("shape:", x.shape)

xT = x.T
print("\nx^T (vector hàng):")
print(xT)
print("shape:", xT.shape)
```

## 4. Chuyển vector hàng thành vector cột

$$
\mathbf{v} = \begin{bmatrix} 4 & 5 & 6 \end{bmatrix}
\quad \xrightarrow{\ T\ } \quad
\mathbf{v}^T = \begin{bmatrix} 4 \ 5 \ 6 \end{bmatrix}
$$

Kích thước đổi từ $1 \times 3$ thành $3 \times 1$.

```python
v = np.array([[4, 5, 6]])
print("v (vector hàng):")
print(v)
print("shape:", v.shape)

vT = v.T
print("\nv^T (vector cột):")
print(vT)
print("shape:", vT.shape)
```

## 5. Chuyển vị hai lần thì được lại vector ban đầu

$(\mathbf{x}^T)^T = \mathbf{x}$: cột → hàng → cột, và hàng → cột → hàng.

```python
print("(x^T)^T == x:", np.array_equal(x.T.T, x))
print("(v^T)^T == v:", np.array_equal(v.T.T, v))
```
