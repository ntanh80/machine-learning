# Cập nhật trọng số trong Gradient Descent

*Bài 5, mục 5.1.3.* Quy tắc cập nhật: $x_{t+1} = x_t - \eta f'(x_t)$ (một biến), $\mathbf{w}_{t+1} = \mathbf{w}_t - \eta\nabla\mathcal{L}(\mathbf{w}_t)$ (nhiều biến).

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Tính tay với $f(x) = x^2 + 4x + 4$

$f'(x) = 2x + 4$, $x_0 = 0$, $\eta = 0.25$. Cực tiểu tại $x^* = -2$.

```python
x = 0.0
for t in range(4):
    g = 2 * x + 4
    x_new = x - 0.25 * g                     # x_{t+1} = x_t - eta * f'(x_t)
    print("t = %d: x = %7.4f, f'(x) = %6.3f -> x mới = %7.4f, f = %.4f"
          % (t, x, g, x_new, (x_new + 2)**2))
    x = x_new
```

## 2. Một bước GD cho hàm mất mát MSE

Dữ liệu 3 điểm ở notebook 01, $\eta = 0.1$. Mọi tham số được cập nhật **đồng thời**.

```python
X = np.array([[1, 1], [1, 2], [1, 3]], dtype=float)
y = np.array([2, 3, 5], dtype=float)
loss = lambda w: np.sum((X @ w - y)**2) / (2 * len(y))

w = np.zeros(2)
g = X.T @ (X @ w - y) / len(y)
w_new = w - 0.1 * g
print('w = %s, L = %.3f  ->  w = %s, L = %.3f' % (w, loss(w), w_new.round(3), loss(w_new)))
```

## 3. Hàm $f(x) = x^2 + 5\sin x$

$f'(x) = 2x + 5\cos x = 0$ không giải được bằng công thức, dùng GD với hai điểm khởi tạo. Thuật toán dừng khi $|f'(x)| < 10^{-3}$.

```python
def f(x):
    # Hàm cần tối thiểu f(x)
    return x**2 + 5 * np.sin(x)


def grad(x):
    # Đạo hàm f'(x)
    return 2 * x + 5 * np.cos(x)


def gd_1d(x0, eta, max_iter=100, tol=1e-3):
    # GD từ x0 với tốc độ học η, dừng khi |f'(x)| < tol; trả về dãy x_t
    x = x0
    history = [x0]
    for t in range(max_iter):
        x = x - eta * grad(x)        # x_{t+1} = x_t - eta * f'(x_t)
        history.append(x)
        if abs(grad(x)) < tol:       # đạo hàm đủ nhỏ -> dừng
            break
    return np.array(history)


x_line = np.linspace(-6, 6, 400)
fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
for ax, x0 in zip(axes, [-5, 5]):
    x_history = gd_1d(x0, eta=0.1)
    print('x0 = %2d: x = %.4f, f(x) = %.4f, sau %d vòng lặp' % (x0, x_history[-1], f(x_history[-1]), len(x_history) - 1))
    ax.plot(x_line, f(x_line), color='gray')
    ax.plot(x_history, f(x_history), 'ro-', ms=4)
    ax.set_title(r'$x_0 = %d$: %d vòng lặp' % (x0, len(x_history) - 1))
plt.show()
```

Từ $x_0 = 5$ thuật toán chậm hơn vì phải đi qua vùng gần như bằng phẳng quanh $x \approx 2.7$ (đạo hàm nhỏ nên bước đi ngắn).
