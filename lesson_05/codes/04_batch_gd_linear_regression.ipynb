# Tối ưu hàm mất mát MSE bằng Batch Gradient Descent

*Bài 5, mục 5.3.1, Bước 1.* 1000 điểm nằm gần đường thẳng $y = 4 + 3x$. Dùng nghiệm OLS để kiểm tra nghiệm của GD.

```python
import numpy as np

np.random.seed(2)
X = np.random.rand(1000)                       # x trong [0, 1)
y = 4 + 3 * X + .2 * np.random.randn(1000)     # y = 4 + 3x + nhiễu
Xbar = np.column_stack([np.ones(len(X)), X])   # mỗi hàng [1, x_i]

w_ols = np.linalg.pinv(Xbar.T @ Xbar) @ Xbar.T @ y
print('OLS: w =', w_ols)
```

```python
def loss(w, Xb, y):
    return np.sum((Xb @ w - y)**2) / (2 * len(y))


def grad(w, Xb, y):
    return Xb.T @ (Xb @ w - y) / len(y)


def batch_gd(Xb, y, w_init, eta, max_iter=10000, tol=1e-4):
    w = w_init.copy()
    history = [loss(w, Xb, y)]
    for t in range(1, max_iter + 1):
        w = w - eta * grad(w, Xb, y)
        history.append(loss(w, Xb, y))
        if np.linalg.norm(grad(w, Xb, y)) < tol:   # gradient đủ nhỏ -> dừng
            break
    return w, t, history
```

## So sánh các tốc độ học

```python
for eta in [1, 0.1]:
    w, t, history = batch_gd(Xbar, y, np.array([2., 1.]), eta)
    print('eta = %-4s w = %s, sau %d vòng lặp, loss = %.6f' % (eta, w.round(4), t, history[-1]))
```

Tốc độ học quá lớn làm GD phân kỳ:

```python
with np.errstate(over='ignore', invalid='ignore'):
    w, t, history = batch_gd(Xbar, y, np.array([2., 1.]), 2.5, max_iter=1000)
print('eta = 2.5  w =', w)
```
