# Vẽ tiếp tuyến và dấu của đạo hàm (mục 5.1.2)

Hàm f(x) = (x - 1)^2 / 2 - 2 với tiếp tuyến tại x = -1, x = 1 và x = 3

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

IMAGES = Path('.')   # thư mục images/ (notebook chạy trong images/)

def f(x):
    return 0.5 * (x - 1)**2 - 2

def grad(x):
    """Đạo hàm f'(x)."""
    return x - 1

x_line = np.linspace(-3, 5, 400)
fig, ax = plt.subplots(figsize=(7.5, 4.6))
ax.plot(x_line, f(x_line), color="#7F7F7F", lw=2, label=r"$f(x) = \frac{1}{2}(x - 1)^2 - 2$")

# Mỗi điểm: hoành độ, màu tiếp tuyến, nội dung chú thích, vị trí chú thích
points = [(-1, "#1F77B4", "$f'(-1) = -2 < 0$\nđi sang phải thì $f$ giảm", (-2.85, -2.6)),
          (1, "#2CA02C", "$f'(1) = 0$\ntiếp tuyến nằm ngang", (0.05, -3.55)),
          (3, "#D62728", "$f'(3) = 2 > 0$\nđi sang trái thì $f$ giảm", (2.75, -2.6))]
for x0, color, text, text_position in points:
    # Tiếp tuyến tại x0: y = f(x0) + f'(x0)(x - x0)
    x_tangent = np.array([x0 - 1.2, x0 + 1.2])
    ax.plot(x_tangent, f(x0) + grad(x0) * (x_tangent - x0), color=color, lw=1.8)
    ax.plot(x0, f(x0), "o", color=color, ms=8)
    ax.annotate(text, xy=(x0, f(x0)), xytext=text_position, fontsize=10)

ax.set_xlim(-3, 5); ax.set_ylim(-4, 5)
ax.set_xlabel("$x$"); ax.set_ylabel("$f(x)$"); ax.grid(alpha=0.3)
ax.legend(loc="upper center", fontsize=10)
fig.tight_layout()
fig.savefig(IMAGES / "derivative_slope.png", dpi=150)
```
