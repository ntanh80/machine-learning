# Thuật toán PLA: ví dụ tính tay

*Bài 4, mục 4.2.3 d).* Bốn điểm A, B (lớp $+1$) và C, D (lớp $-1$), khởi tạo $\mathbf{w} = [w_0, w_1, w_2] = [0, 0, 0]$, duyệt các điểm theo thứ tự. Gặp điểm sai thì cập nhật $\mathbf{w} \leftarrow \mathbf{w} + y_i\mathbf{x}_i$.

```python
import numpy as np

names = ['A', 'B', 'C', 'D']
X = np.array([[1, 1], [2, 1], [1, -1], [-1, -2]])
y = np.array([1, 1, -1, -1])
Xbar = np.column_stack([np.ones(len(X)), X])   # vector mở rộng [1, x1, x2]


def sgn(z):
    return 1 if z >= 0 else -1                 # quy ước sgn(0) = 1
```

## In từng bước của thuật toán

```python
w = np.zeros(3)
epoch = 0
while True:
    epoch += 1
    updated = False
    print('Vòng %d:' % epoch)
    for i in range(len(y)):
        z = Xbar[i] @ w
        if sgn(z) != y[i]:
            w = w + y[i] * Xbar[i]             # w <- w + y_i x_i
            updated = True
            print('  %s: z = %5.1f -> SAI, cập nhật w = %s' % (names[i], z, w))
        else:
            print('  %s: z = %5.1f -> đúng' % (names[i], z))
    if not updated:
        break

print('Nghiệm w =', w, '-> siêu phẳng %.0f + %.0f x1 + %.0f x2 = 0' % tuple(w))
```
